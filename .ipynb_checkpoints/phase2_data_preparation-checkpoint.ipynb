## Why each preprocessing step is necessary

**1. Lowercase conversion**
Ensures the model treats "Shipment" and "shipment" as the same word.
Without this, the same word in different cases would be counted as different features.

**2. Punctuation and number removal**
Punctuation carries no meaning for classification.
Numbers are removed because the only numeric references in this dataset
are shift numbers (e.g. "shift 3"), which do not indicate incident category.

**3. Tokenization**
Splits each sentence into individual words so they can be processed one at a time.
This is a required step before stopword removal and lemmatization can work.

**4. Stopword removal**
Common words like "the", "is", "at", "during", "reported" appear in every ticket
regardless of category. Removing them reduces noise and helps the model focus
on meaningful words like "damaged", "delayed", or "customs".

**5. Lemmatization**
Reduces words to their base form. "delayed" becomes "delay", "damaged" becomes "damage".
This prevents the model from treating the same concept as different features
just because of tense or plural form.

```python
df = pd.read_csv("supply_chain_incidents_v2_5000.csv")
```

<b> Note: </b>The descriptions are short and templated, so preprocessing removes a significant portion of each sentence (words like 'reported', 'facility', and 'shift') appear in almost every ticket regardless of category and will likely be removed or downweighted as stopwords. Lemmatization has limited impact here since the text is already quite simple.

```python
df[['incident_id', 'incident_category', 'severity' , 'incident_description', 'resolution_status' , 'location',  'clean_text']].to_csv("clean_data.csv", index=False)
print("Saved.")
```
