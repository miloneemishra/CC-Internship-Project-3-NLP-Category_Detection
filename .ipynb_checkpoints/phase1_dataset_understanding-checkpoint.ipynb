```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

# PHASE 1 - Dataset Understanding

```python
df = pd.read_csv("supply_chain_incidents_v2_5000.csv")
print(df.shape)
```

```python
df.head()
```

```python
df.dtypes
```

```python
df.isnull().sum()
```

```python
df['incident_category'].value_counts()
```

```python
df['incident_category'].value_counts().plot(kind='bar', color='lightblue', edgecolor='black')
plt.title('Incident Category Distribution')
plt.xlabel('Category')
plt.ylabel('Count')
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

```python
df['severity'].value_counts().plot(kind='bar', color='lightblue', edgecolor='black')
plt.title('Severity Distribution')
plt.xlabel('Severity')
plt.ylabel('Count')
plt.tight_layout()
plt.show()
```

```python
df['resolution_status'].value_counts().plot(kind='bar', color='lightblue', edgecolor='black')
plt.title('Resolution Status Distribution')
plt.xlabel('Status')
plt.ylabel('Count')
plt.tight_layout()
plt.show()
```

<b>Key Observations:</b> The dataset contains 5000 rows and 6 columns with no missing values. All 7 categories are well balanced (~700 tickets each), which is ideal for training a fair classifier. Incident descriptions follow a templated structure, which may lead to higher than expected model accuracy in later phases.
